## Task 1 - A line chart showing CO2 per Capita for all countries in the dataset

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country/countries **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country/countries and its story


```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# 1. Load dataset (adjust path as needed)
df = pd.read_csv('../data/co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")

# Define configuration for the highlighted country
highlight_country = 'China'
highlight_color = '#ef4444' # Vibrant red for emphasis

fig = go.Figure()

# 2. Add background context lines for all other countries (thin & light grey #EEEEEE)
for country in df['Country'].unique():
    if country != highlight_country:
        cdf = df[df['Country'] == country]
        fig.add_trace(go.Scatter(
            x=cdf['Year'],
            y=cdf['CO2_per_capita'],  # Replace with 'PerCapita_CO2' if matching your CSV schema
            mode='lines',
            line=dict(color='#EEEEEE', width=1),
            hoverinfo='skip',
            showlegend=False
        ))

# 3. Add the highlighted country line (thicker, distinct color)
h_df = df[df['Country'] == highlight_country].sort_values('Year')
fig.add_trace(go.Scatter(
    x=h_df['Year'],
    y=h_df['CO2_per_capita'],
    mode='lines',
    name=highlight_country,
    line=dict(color=highlight_color, width=3),
    showlegend=False
))

# 4. Direct label at the end of the highlighted line (instead of a legend)
last_row = h_df.iloc[-1]
fig.add_annotation(
    x=last_row['Year'],
    y=last_row['CO2_per_capita'],
    text=f"  <b>{highlight_country}</b>",
    xanchor='left',
    yanchor='middle',
    showarrow=False,
    font=dict(color=highlight_color, size=13, family='Inter, sans-serif')
)

# 5. Layout styling with an insight-driven title and extended right margin for the label
fig.update_layout(
    title=dict(
        text="<b>China's Rapid Per Capita Emissions Growth</b>: Surpassing Global Averages Over Two Decades",
        font=dict(size=15, family='Inter, sans-serif')
    ),
    xaxis_title="Year",
    yaxis_title="CO2 Emissions per Capita (Tonnes)",
    template="plotly_white",
    width=950,
    height=600,
    margin=dict(r=120, t=80, l=80, b=80), # Extra right margin to display the direct label cleanly
    font=dict(family='Inter, sans-serif', size=12)
)

fig.show()
```

## Task 2 — A line chart showing CO2 emissions over time for all Asian countries in the dataset, with one country highlighted.
